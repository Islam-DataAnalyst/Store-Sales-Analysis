```python
# Block 1: Importing libraries
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python

# Block 2: Loading data and initial inspection
# Load the dataset
df = pd.read_csv('Superstore Sales Dataset.csv')

print(f"Dataset Shape: {df.shape[0]} rows and {df.shape[1]} columns.\n")
print("--- Data Types and Overview ---")
print(df.info())

print("\n--- Summary of Missing (Null) Values ---")
print(df.isnull().sum()[df.isnull().sum() > 0])

print("\n--- Summary of Duplicate Rows ---")
print(f"Number of duplicate rows: {df.duplicated().sum()}")
```

```python
# Block 3: Data Cleaning

# 1. Fill missing Postal Codes for Burlington, Vermont (ZIP: 05401)
df['Postal Code'] = df['Postal Code'].fillna(5401)

# 2. Convert Postal Code to integer type (removes the trailing .0)
df['Postal Code'] = df['Postal Code'].astype(int)

# 3. Convert date columns to datetime objects (handling day/month/year format)
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%d/%m/%Y')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], format='%d/%m/%Y')

print("Data cleaning completed successfully!")
print("\nDouble checking missing values:")
print(df.isnull().sum())
print("\nUpdated column data types:")
print(df[['Order Date', 'Ship Date', 'Postal Code']].dtypes)
```

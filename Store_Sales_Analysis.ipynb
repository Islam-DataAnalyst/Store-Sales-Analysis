```python
# Block 1: Importing Libraries and Loading Data
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv('Superstore Sales Dataset (1).csv')

# Convert dates to datetime objects using the correct format (Day/Month/Year)
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%d/%m/%Y')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], format='%d/%m/%Y')

print("Libraries imported and data loaded successfully.")
print(f"Dataset Shape: {df.shape[0]} rows and {df.shape[1]} columns.")
```

```python
# Block 2: Executive KPIs
total_sales = df['Sales'].sum()
total_orders = df['Order ID'].nunique()
aov = total_sales / total_orders
avg_sales_per_customer = df.groupby('Customer ID')['Sales'].sum().mean()

print("--- Executive KPIs ---")
print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Orders: {total_orders}")
print(f"Average Order Value (AOV): ${aov:,.2f}")
print(f"Avg Sales per Customer: ${avg_sales_per_customer:,.2f}")
```

```python
# Block 3: Category Breakdown & Pareto Analysis
print("--- Sales Contribution by Category (%) ---")
cat_sales_pct = (df.groupby('Category')['Sales'].sum() / total_sales * 100).round(2)
print(cat_sales_pct.sort_values(ascending=False))

# Pareto Analysis (80/20 Rule)
# Sort products by sales descending
prod_sales = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False)

# Calculate cumulative percentage
cum_sales = prod_sales.cumsum() / total_sales

# Find what percentage of sales the Top 20% of products generate
top_20_percent_cutoff = int(len(prod_sales) * 0.2)
pareto_sales_pct = cum_sales.iloc[top_20_percent_cutoff - 1] * 100

print(f"\n--- Pareto Principle Check ---")
print(f"The Top 20% of our products generate {pareto_sales_pct:.2f}% of our total sales.")
```

```python
# Block 4: Sales by Segment
segment_sales = df.groupby('Segment')['Sales'].sum().sort_values(ascending=False)

print("--- Total Sales by Segment ---")
# Formatting the output for readability
for segment, sales in segment_sales.items():
    print(f"{segment}: ${sales:,.2f}")
```

```python
# Block 5: Regional and State Performance
region_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)

print("--- Sales by Region ---")
for region, sales in region_sales.items():
    print(f"{region}: ${sales:,.2f}")

print("\n--- Top 5 States by Sales ---")
top_states = df.groupby('State')['Sales'].sum().nlargest(5)
print(top_states.apply(lambda x: f"${x:,.2f}"))
```

```python
# Block 6: Logistics and Shipping Speed
# Feature Engineering: Calculate delivery duration in days
df['Shipping Time (Days)'] = (df['Ship Date'] - df['Order Date']).dt.days

avg_shipping = df['Shipping Time (Days)'].mean()
print(f"Overall Average Shipping Time: {avg_shipping:.2f} days\n")

print("--- Average Shipping Time by Mode (Days) ---")
shipping_modes = df.groupby('Ship Mode')['Shipping Time (Days)'].mean().sort_values()
print(shipping_modes.round(2))
```
